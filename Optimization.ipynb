```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
from itertools import combinations

# Load data
file_path = 'HIT140_World_Cup_2026_FINAL.xlsx'
df_2_1 = pd.read_excel(file_path, sheet_name='Objective_2.1')
df_2_2 = pd.read_excel(file_path, sheet_name='Objective_2.2')

print("✓ Data loaded")
print(f"  Objective 2.1: {df_2_1.shape}")
print(f"  Objective 2.2: {df_2_2.shape}")
```

```python
print("="*70)
print("OBJECTIVE 2.1: GOAL DIFFERENCE - FEATURE OPTIMIZATION")
print("="*70)

all_pred_2_1 = ['Elo_Ranking_Difference', 'FIFA_Ranking_Difference', 
                 'Avg_Goals_Scored_Difference', 'Avg_Goals_Conceded_Difference',
                 'Head_to_Head_Record', 'Tournament_Stage',
                 'Squad_Elo_Interaction', 'Goals_Form_Interaction']

y_2_1 = df_2_1['Goal_Difference']
best_r2_2_1 = -999
best_feat_2_1 = []
best_model_2_1 = None
results_2_1 = []

# Test all feature combinations
for num_feat in range(3, 9):
    print(f"  Testing {num_feat} features...")
    for combo in combinations(all_pred_2_1, num_feat):
        X = df_2_1[list(combo)]
        model = LinearRegression().fit(X, y_2_1)
        y_pred = model.predict(X)
        r2 = r2_score(y_2_1, y_pred)
        rmse = np.sqrt(mean_squared_error(y_2_1, y_pred))
        
        results_2_1.append({'features': combo, 'r2': r2, 'rmse': rmse})
        
        if r2 > best_r2_2_1:
            best_r2_2_1 = r2
            best_feat_2_1 = combo
            best_model_2_1 = model

# Sort and show top 10
results_2_1.sort(key=lambda x: x['r2'], reverse=True)

print(f"\nTop 10 Combinations:")
print(f"{'Rank':<5} {'R²':<8} {'RMSE':<8} {'# Features':<12}")
print("-"*35)
for i, r in enumerate(results_2_1[:10], 1):
    print(f"{i:<5} {r['r2']:<8.4f} {r['rmse']:<8.4f} {len(r['features']):<12}")

print(f"\n{'='*70}")
print(f"BEST MODEL - Objective 2.1")
print(f"{'='*70}")
print(f"R² Score: {best_r2_2_1:.4f}")
print(f"Number of Features: {len(best_feat_2_1)}")
print(f"\nFeatures Used:")
for feat in best_feat_2_1:
    print(f"  - {feat}")
print(f"\nCoefficients:")
for feat, coef in zip(best_feat_2_1, best_model_2_1.coef_):
    print(f"  {feat}: {coef:.6f}")
print(f"Intercept: {best_model_2_1.intercept_:.4f}")
print(f"\nImprovement over baseline (0.449): {best_r2_2_1 - 0.449:+.4f}")
```

```python
from sklearn.preprocessing import PolynomialFeatures

print("="*70)
print("OBJECTIVE 2.1: POLYNOMIAL FEATURES (DEGREE 2)")
print("="*70)

all_pred_2_1 = ['Elo_Ranking_Difference', 'FIFA_Ranking_Difference', 
                 'Avg_Goals_Scored_Difference', 'Avg_Goals_Conceded_Difference',
                 'Head_to_Head_Record', 'Tournament_Stage',
                 'Squad_Elo_Interaction', 'Goals_Form_Interaction']

X_2_1_base = df_2_1[all_pred_2_1]
y_2_1 = df_2_1['Goal_Difference']

# Create polynomial features (includes original + squared terms)
poly = PolynomialFeatures(degree=2, include_bias=False)
X_2_1_poly = poly.fit_transform(X_2_1_base)

# Train model
model_poly_2_1 = LinearRegression().fit(X_2_1_poly, y_2_1)
y_pred_poly_2_1 = model_poly_2_1.predict(X_2_1_poly)

r2_poly_2_1 = r2_score(y_2_1, y_pred_poly_2_1)
rmse_poly_2_1 = np.sqrt(mean_squared_error(y_2_1, y_pred_poly_2_1))

print(f"\nObjective 2.1 - Polynomial Features:")
print(f"  Original features: 8")
print(f"  Polynomial features: {X_2_1_poly.shape[1]}")
print(f"  R² Score: {r2_poly_2_1:.4f}")
print(f"  RMSE: {rmse_poly_2_1:.4f}")
print(f"  vs Baseline (0.449): {r2_poly_2_1 - 0.449:+.4f}")

# OBJECTIVE 2.2
print("\n" + "="*70)
print("OBJECTIVE 2.2: POLYNOMIAL FEATURES (DEGREE 2)")
print("="*70)

all_pred_2_2 = ['Elo_Rating', 'FIFA_Ranking', 'Avg_Goals_Scored', 
                 'Avg_Goals_Conceded', 'Squad_Value', 'Opponent_FIFA_Ranking',
                 'Opponent_Avg_Goals_Conceded', 'Team_Rest_Days']

X_2_2_base = df_2_2[all_pred_2_2]
y_2_2 = df_2_2['Goals_Scored']

poly_2_2 = PolynomialFeatures(degree=2, include_bias=False)
X_2_2_poly = poly_2_2.fit_transform(X_2_2_base)

model_poly_2_2 = LinearRegression().fit(X_2_2_poly, y_2_2)
y_pred_poly_2_2 = model_poly_2_2.predict(X_2_2_poly)

r2_poly_2_2 = r2_score(y_2_2, y_pred_poly_2_2)
rmse_poly_2_2 = np.sqrt(mean_squared_error(y_2_2, y_pred_poly_2_2))

print(f"\nObjective 2.2 - Polynomial Features:")
print(f"  Original features: 8")
print(f"  Polynomial features: {X_2_2_poly.shape[1]}")
print(f"  R² Score: {r2_poly_2_2:.4f}")
print(f"  RMSE: {rmse_poly_2_2:.4f}")
print(f"  vs Baseline (0.287): {r2_poly_2_2 - 0.287:+.4f}")
```
